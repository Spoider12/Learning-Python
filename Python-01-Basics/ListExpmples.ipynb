```python
grades = [85,25,65,75,87]

grades.append(95) ##Adding a new Grade

## Calculating the average grades
average_grade = sum(grades) / len(grades)
print(f"Avergae Grade :{average_grade}")

##Calculating who has highest grades and who has lowest grades
highest_grade = max(grades)
lowest_grades = min(grades)

print(f"Highest Grade:{highest_grade}")
print(f"Lowest Grade:{lowest_grades}")
```
