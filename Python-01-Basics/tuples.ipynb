```python
##Packing and unpacking tuple
packing_tuple = "Hello",1,3.14,"Krish"
```

```python
##packing tuple
print(packing_tuple)
```

```python
##unpacking tuple
a,b,c,d = packing_tuple
print(a)
print(b)
print(c)
print(d)
```

```python
##unpacking with **
numbers =(1,2,3,4,5,6,7)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```
