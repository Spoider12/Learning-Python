```python
## Single key is always used
student ={"name": "anurag","age": 23,"grade": 'A'}
print(student)
print(type(student))
```

```python
##Accessing Dictionary Elements
print(student['grade'])
print(student['age'])
```

```python
## Accessing using get()method provided by dictionary
print(student.get('grade'))
print(student.get('last_name'))
```

```python
print(student.get('last_name',"Not Available"))
```

```python
##Modifying Dictionary Elemets
## Dictionary are mutable so you can add update or delete element
print(student)
```

```python
student['age']=33 ##update value the key 
print(student)
student["address"]="India" ## adding a new key
print(student)
```

```python
del student ["grade"] ## delete key and value pair
print(student)
```

```python
keys=student.keys() ##get all the keys
print(keys)
values=student.values() ##get all the values
print(values)
items =student.items()
print(items) ##get all key value pairs
```

```python
## shallow copy
student_copy=student
print(student)
print(student_copy)
```

```python
student["name"] = "Krish2"
print(student)
print(student_copy)
```

```python
### iterating over dictionaries
## you can use loops to iterate over dictionaries , keys,values,or items
for keys in student.keys():
    print(keys)
```

```python
for value in student.values():
    print(value)
```

```python
## iterate over key value pairs
for key,value in student.items():
    print(f"{key}:{value}")
```

```python
## Nested deictionaries
student ={
    "student1" : {"name":"Anurag","age":23},
    "student2" : {"name":"Anjali","age":25}
}
print(student)
```

```python
## accessing Nested decitonaries elements
print(student["student2"]["name"])
print(student["student2"]["age"])
```

```python
student.items()
```

```python
for student_id,student_info in student.items():
    print(f"{student_id}:{student_info}")
    for key ,value in student_info.items():
        print(f"{key}:{value}")
```

```python
sqaure ={x:x**2 for x in range(10)}
print(sqaure)
```

```python
evens ={x:x**2 for x in range(5) if x%2==0}
print(evens)
```

```python
numbers = [1,2,3,4,5,62,3,4,5,2,3,2,5,4,3,6,5,3]
frequency ={}
for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1
print(frequency)
```

```python
## Merge 2 Dictionaries into one
dict_1 ={"a":1,"b":2}
dict_2 ={"b":3,"c":4}
merged_dict={**dict_1,**dict_2}
print(merged_dict)
```
