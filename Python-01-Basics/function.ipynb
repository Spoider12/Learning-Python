```python

def even_or_odd(num):
    """Identify the number is even or odd"""
    if num%2==0:
        print("The Number is even")
    else:
        print("The Number is odd")
```

```python
##Variable length Arguments
## Positional and Keywords argument
##positional arguments
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,67,8,9,0, "anurag")
```

```python
## keywords arguments
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_details(name="Anurag", age="23",country="India")
```

```python
##Combine both
def print_value(*args, **kwargs):
    for val in args:
        print(f" Positional Parameters{val}")
    for key,value in kwargs.items():
        print(f" Keyword Arguments{key}:{value}")
```

```python
print_value(1,2,3,4,5,"Krish",name="Anurag", age="23",country="India")
```
