```python
## Read a Whole File
with open('example.txt','r') as file:
    content = file.read()
    print(content)
    
      
        
```

```python
with open('example.txt','r') as file:
    for line in file:
        print(line.strip()) ## strip() removes the new line character
```

```python
## writing a file overwriting

with open('example.txt','w') as file:
    file.write("Also worked on Several projects \n")
    file.write("that builds my problem solving skills and technical skills\n")
```

```python
## write a file without overwriting
with open('example.txt','a') as file:
    file.write("Do DSA To imporve my problem solving skill continuosly\n")
```

```python
### Writing a list of lines to a file 
lines =['First Line\n','Second Line \n', 'Third Line \n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
## Binary Files 
# Writing a binary file 
data = b'\x00\x01\x02\x03\x04'
with open('exmaple.bin','wb')as file:
    file.write(data)
  
```

```python
## read a binary file 
with open('exmaple.bin','rb')as file:
    content = file.read()
    print(content)
```

```python
### Read the content from source file and wirte to the destination file
# Copying the text file
with open('example.txt','r')as source_file:
    content = source_file.read()
    
with open('destination.txt','w')as destination_file:
    destination_file.write(content)
```

```python
## Read a text files and count the number of lines , words , and characters
with open('example.txt','r') as file:
    content =file.read()
    lines =len(content.splitlines())
    words =len(content.split())
    characters =len(content)
    
    print("Lines:",lines)
    print("Words:",words)
    print("Characters:",characters)
        
```

### the w+ mode in python is used to open a file for both reading and writing. If the file does not exist , it will be created. if the file exists, its content is truncated(i.e.., the file is overwritten)

```python
## writing a reading a file 

with open('example.txt','w+')as file:
    file.write("Hello Anushka\n")
    file.write("Are You Okay <3 \n")
    
    ## Move the file cursor to the beginning 
    file.seek(0)
    ## read the content of the file 
    content = file.read()
    print(content)
```
